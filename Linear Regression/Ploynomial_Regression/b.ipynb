```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
x=6*np.random.rand(100,1)-3
y=0.5 * x**2+1.5*x+2+np.random.randn(100,1)
```

```python
plt.scatter(x,y,color='teal',marker='*')
plt.xlabel('X dataset')
plt.ylabel('Y dataset')
```

```python
print(x.shape)
print(y.shape)
```

```python
## try to apply simple linear regression to check whether it will work or not
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.25,random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
model=LinearRegression()
model.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
```

```python
#calculate r2 score
from sklearn.metrics import r2_score
score=r2_score(y_test,y_pred)
print("R2 Score of my model is :",score)
```

```python
##check for best fit line with non_linear data
plt.scatter(x_train,y_train,marker='*',color='teal')
plt.plot(x_train,model.predict(x_train),color='red')
```

```python
## apply polynomial regression technique
from sklearn.preprocessing import PolynomialFeatures
poly=PolynomialFeatures()
x_train_poly=poly.fit_transform(x_train)
x_test_poly=poly.transform(x_test)
```

```python
##after applying the polynomial technique again check for the regression acccuracy
model.fit(x_train_poly,y_train)
y_predic=model.predict(x_test_poly)
```

```python
#calculate r2 score 
from sklearn.metrics import r2_score
score=r2_score(y_test,y_predic)
print("R2 Score of my model is :",score)
```

here we can see that my accuracy has increased 


```python
plt.scatter(x_train,y_train,color='teal',marker='*')
plt.plot(x_train,model.predict(x_train_poly),color='black')
```

here we can see that after applying the polynomial tech we somehow capable to shift the best fit line to fit in non_linear data and we get better accuracy
