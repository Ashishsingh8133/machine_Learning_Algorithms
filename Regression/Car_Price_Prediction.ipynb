```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
import warnings
warnings.filterwarnings('ignore')

%matplotlib inline
```

```python
df=pd.read_csv('cardekho_imputated.csv')
```

```python
df.head()
```

```python
##drop unnecessory features
df.drop(columns='Unnamed: 0',inplace=True)
```

```python
##check for null values
df.isnull().sum()
```

```python
##extracting features 
## extracting numerical features
numerical_features=[features for features in df.columns if df[features].dtypes!='str']
print(numerical_features,"\n number of numerical features are:",len(numerical_features))
```

```python
##extracting categorical features
categorical_features=[features for features in df.columns if df[features].dtypes=='str']
print(categorical_features,"\n number of categorical features are:",len(categorical_features))
```

```python
##check for discrete numerical features 
discrete_num_features=[features for features in numerical_features if len(df[features].unique())<=25]
print(discrete_num_features,'\n Number of discrete numerical features are:',len(discrete_num_features))
```

```python
##check for continous numerical features 
continous_num_features=[features for features in numerical_features if len(df[features].unique())>=25]
print(continous_num_features,'\n number of continous numerical features:',len(continous_num_features))
```

```python
print(df['vehicle_age'].unique())
print(df['seats'].unique())
print(df['vehicle_age'].value_counts(normalize=True)*100)
print(df['seats'].value_counts(normalize=True)*100,2)
```

```python

##dropping columns
df.drop(columns=['car_name','brand'],inplace=True)
```

```python
df.head(5)
```

```python
##independent & dependent features
x=df.iloc[:,:-1]
y=df.iloc[:,-1]
```

```python
y.head()
```

```python
x.head()
```

```python
#perform label encoding over model feature 
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()

x['model']=le.fit_transform(x['model'])
```

```python
##perform train test split
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.20,random_state=42)
```

```python
pd.DataFrame(x_train)
```

```python
pd.DataFrame(x_test)
```

```python
##create transformers to transform numerical and categorical features
num_features=x.select_dtypes(exclude='str').columns
cat_features=['seller_type','fuel_type','transmission_type']

##import transformers
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import StandardScaler
from sklearn.compose import ColumnTransformer

##load transformers
numerical_transformer=StandardScaler()
categorical_transformer=OneHotEncoder(drop='first')

preprocessor=ColumnTransformer(
    [
        ("OneHotEncoder",categorical_transformer,cat_features),
        ("Standard Scaler",numerical_transformer,num_features)
    ])

```

```python
x_train=preprocessor.fit_transform(x_train)
x_test=preprocessor.transform(x_test)
```

```python
##import regression models 
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.neighbors import KNeighborsRegressor
from xgboost import XGBRegressor

##import performance matrices 
from sklearn.metrics import r2_score,mean_absolute_error,mean_squared_error,root_mean_squared_error
```

```python
models={
    'Decision Tree':DecisionTreeRegressor(),
    'Random Forest':RandomForestRegressor(),
    'Gradient Boost':GradientBoostingRegressor(),
    'K-neighbours':KNeighborsRegressor(),
    'xgboost Regressor':XGBRegressor()
}

for i in range(len(list(models))): ##iterating over models
    model=list(models.values())[i] ##load the models
    model.fit(x_train,y_train) ##fit model on x_train and y_train

    ##make predictions
    y_train_pred=model.predict(x_train)
    y_test_pred=model.predict(x_test)

    ##training set performance matrices 
    model_train_r2=r2_score(y_train,y_train_pred)
    model_train_mean_squared=mean_squared_error(y_train,y_train_pred)
    model_train_mean_absolute=mean_absolute_error(y_train,y_train_pred)
    model_train_root_mean=root_mean_squared_error(y_train,y_train_pred)

    ##testing set performance metrices
    model_test_r2=r2_score(y_test,y_test_pred)
    model_test_mean_squared=mean_squared_error(y_test,y_test_pred)
    model_test_mean_absolute=mean_absolute_error(y_test,y_test_pred)
    model_test_root_mean=root_mean_squared_error(y_test,y_test_pred)

    print(list(models.keys())[i])

    print('------Trainig Set performance------')

    print("-R2 Score {:.4f}".format(model_train_r2))
    print("-Mean Squared Error {:.4f}".format(model_train_mean_squared))
    print("-Mean Absolute Error {:.4f}".format(model_train_mean_absolute))
    print("-Root Mean Squared Error {:.4f}".format(model_train_root_mean))

    print('--------Testing set performance------')

    print("-R2 Score {:.4f}".format(model_test_r2))
    print("-Mean Square Error {:.4f}".format(model_test_mean_squared))
    print("-Mean Absolute Error {:.4f}".format(model_test_mean_absolute))
    print("-Root Mean Squared Error {:.4f}".format(model_test_root_mean))

    print("="*30)
    print("\n")






    
```

```python
##hyperparameter tunning 
from sklearn.model_selection import RandomizedSearchCV
from sklearn.tree import DecisionTreeRegressor

dt_params={
    'criterion':['squared_error','friedman_mse','absolute_error','poisson'],
    'splitter':['best','random'],
    'max_depth':[2,4,5,6,8],
    'min_samples_split':[2,5,10,12],
    'max_features':['auto','sqrt','log2']

}

model_params={}

random_cv=[
    ('Decision Tree Regressor',DecisionTreeRegressor(),dt_params)
]

for name,model,params in random_cv:
    model=RandomizedSearchCV(estimator=model,param_distributions=params,cv=5,refit=True,verbose=2,n_iter=100,n_jobs=-1)
    model.fit(x_train,y_train)
    model_params[name]=model.best_params_

for model_name in model_params:
    print(f'------Best Parameters for {model_name}---------')
    print(model_params[model_name])
    
```

```python
models={
    'Decision Tree':DecisionTreeRegressor(splitter='best',min_samples_split=5,max_features='log2',max_depth=8,criterion='squared_error'),
}

for i in range(len(list(models))): ##iterating over models
    model=list(models.values())[i] ##load the models
    model.fit(x_train,y_train) ##fit model on x_train and y_train

    ##make predictions
    y_train_pred=model.predict(x_train)
    y_test_pred=model.predict(x_test)

    ##training set performance matrices 
    model_train_r2=r2_score(y_train,y_train_pred)
    model_train_mean_squared=mean_squared_error(y_train,y_train_pred)
    model_train_mean_absolute=mean_absolute_error(y_train,y_train_pred)
    model_train_root_mean=root_mean_squared_error(y_train,y_train_pred)

    ##testing set performance metrices
    model_test_r2=r2_score(y_test,y_test_pred)
    model_test_mean_squared=mean_squared_error(y_test,y_test_pred)
    model_test_mean_absolute=mean_absolute_error(y_test,y_test_pred)
    model_test_root_mean=root_mean_squared_error(y_test,y_test_pred)

    print(list(models.keys())[i])

    print('------Trainig Set performance------')

    print("-R2 Score {:.4f}".format(model_train_r2))
    print("-Mean Squared Error {:.4f}".format(model_train_mean_squared))
    print("-Mean Absolute Error {:.4f}".format(model_train_mean_absolute))
    print("-Root Mean Squared Error {:.4f}".format(model_train_root_mean))

    print('--------Testing set performance------')

    print("-R2 Score {:.4f}".format(model_test_r2))
    print("-Mean Square Error {:.4f}".format(model_test_mean_squared))
    print("-Mean Absolute Error {:.4f}".format(model_test_mean_absolute))
    print("-Root Mean Squared Error {:.4f}".format(model_test_root_mean))

    print("="*30)
    print("\n")
```
