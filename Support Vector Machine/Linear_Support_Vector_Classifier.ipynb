```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
from sklearn.datasets import make_classification
```

```python
x,y=make_classification(n_classes=2,n_features=2,n_clusters_per_class=1,n_redundant=0,n_samples=1000)
```

```python
x
```

```python
y
```

```python
pd.DataFrame(x)[1]
```

```python
sns.scatterplot(x=pd.DataFrame(x)[0],y=pd.DataFrame(x)[1],hue=y)
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.20,random_state=42)
```

```python
from sklearn.svm import SVC
model=SVC()
model.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
```

```python
from sklearn.metrics import classification_report,confusion_matrix
print(classification_report(y_test,y_pred))
print(confusion_matrix(y_test,y_pred))
```

```python
##using different SVM Kernel
rbf=SVC(kernel='rbf')
rbf.fit(x_train,y_train)
```

```python
rbf.intercept_
```

```python
y_pred1=rbf.predict(x_test)
```

```python
from sklearn.metrics import classification_report,confusion_matrix
print(classification_report(y_test,y_pred1))
print(confusion_matrix(y_test,y_pred1))
```

```python
##using different kernel
poly=SVC(kernel='poly')
poly.fit(x_train,y_train)
```

```python
y_pred2=poly.predict(x_test)
```

```python
from sklearn.metrics import classification_report,confusion_matrix
print(classification_report(y_test,y_pred2))
print(confusion_matrix(y_test,y_pred2))
```

```python
#using different SVM Kernel
sigmoid=SVC(kernel='sigmoid')
sigmoid.fit(x_train,y_train)
```

```python
y_pred3=sigmoid.predict(x_test)
```

```python
from sklearn.metrics import classification_report,confusion_matrix
print(classification_report(y_test,y_pred3))
print(confusion_matrix(y_test,y_pred3))
```

```python
##Hyperparameter Tunning using GridSearch CV
kernel=['rbf','linear','poly','sigmoid']
gamma=[1,0.1,0.01,0.001,0.0001]
c_values=[0.1,1,10,100,1000]
param=dict(kernel=kernel,C=c_values,gamma=gamma)
print(param)
```

```python
from sklearn.model_selection import GridSearchCV
grid=GridSearchCV(estimator=model,param_grid=param,refit=True,cv=5,verbose=2)
```

```python
grid.fit(x_train,y_train)
```

```python
grid.best_params_
```

```python
grid.best_estimator_
```

```python
grid.best_score_
```

```python
y_new=grid.predict(x_test)
```

```python
from sklearn.metrics import classification_report,confusion_matrix
print(classification_report(y_test,y_new))
print(confusion_matrix(y_test,y_new))
```
