```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.30,random_state=42)
```

```python
##apply label encoding 
from sklearn.preprocessing import LabelEncoder
le1=LabelEncoder()
le2=LabelEncoder()
le3=LabelEncoder()
```

```python
x_train['sex']=le1.fit_transform(x_train['sex'])
x_train['smoker']=le2.fit_transform(x_train['smoker'])
x_train['time']=le3.fit_transform(x_train['time'])
```

```python
x_test['sex']=le1.transform(x_test['sex'])
x_test['smoker']=le2.transform(x_test['smoker'])
x_test['time']=le3.transform(x_test['time'])
```

```python
from sklearn.preprocessing import OneHotEncoder
h1=OneHotEncoder()
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
```

```python
ct=ColumnTransformer(transformers=[('onehot',OneHotEncoder(drop='first'),[3])],remainder='passthrough')
```

```python
x_train=ct.fit_transform(x_train)
```

```python
import sys

import numpy
numpy.set_printoptions(threshold=sys.maxsize)
x_train
```

```python
x_test=ct.transform(x_test)
```

```python
##apply support vector regressor 
from sklearn.svm import SVR
model=SVR()
model.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
```

```python
from sklearn.metrics import r2_score
score=r2_score(y_test,y_pred)
print(score)
```

```python
##Hyper parameter tunning using Grid Search CV
kernel=['rbf']
c_values=[0.1,1,10,100]
gamma=[0.0001,0.001,0.01,0.1,1]
param=dict(kernel=kernel,C=c_values,gamma=gamma)
```

```python
from sklearn.model_selection import GridSearchCV
grid=GridSearchCV(estimator=model,param_grid=param,refit=True,cv=5,verbose=3)

```

```python
grid.fit(x_train,y_train)
```

```python
grid.best_params_
```

```python
y_new=grid.predict(x_test)
```

```python
from sklearn.metrics import r2_score
score=r2_score(y_test,y_new)
print(score)
```
